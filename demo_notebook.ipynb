# RootAI Semantic Parser Demo

Run a bug-bounty focused scan and render Markdown findings.

```python
from rootai_semantic_parser.legacy_impl import AnalysisOptions, MultiFileParser, SecurityConfig, TaintConfig, build_bounty_report

root = '/path/to/repo'
parser = MultiFileParser(root, security_config=SecurityConfig.default_bugbounty(), options=AnalysisOptions(profile='bugbounty', quick_mode=True))
report = parser.scan(taint_config=TaintConfig.profile('bugbounty'))
bounty = build_bounty_report(report, profile='bugbounty', quick_mode=True)
bounty.findings[:3]
```
